```python
def initialize_models(device):
    mtcnn = MTCNN(image_size=160, margin=20, device=device)
    embedder = InceptionResnetV1(pretrained="vggface2").to(device).eval()
    return mtcnn, embedder
```

```python
def build_database(data_root, mtcnn, embedder, device):
    database = {}

    for person_name in os.listdir(data_root):
        person_dir = os.path.join(data_root, person_name)
        if not os.path.isdir(person_dir):
            continue

        embeddings = []

        for img_name in os.listdir(person_dir):
            img_path = os.path.join(person_dir, img_name)
            img = Image.open(img_path).convert("RGB")

            face = mtcnn(img)
            if face is None:
                print(f"⚠️ No face detected in {img_path}")
                continue

            face = face.unsqueeze(0).to(device)

            with torch.no_grad():
                emb = embedder(face)
                embeddings.append(emb.cpu().numpy())

        if len(embeddings) > 0:
            mean_embedding = np.mean(embeddings, axis=0)
            database[person_name] = mean_embedding
            print(f"✅ Saved embedding for {person_name}")

    return database
```

```python
def load_or_create_database(mtcnn, embedder, device):
    if os.path.exists(DB_PATH):
        with open(DB_PATH, "rb") as f:
            database = pickle.load(f)
        print("✅ Loaded authorized embeddings from disk.")
    else:
        database = build_database(DATA_ROOT, mtcnn, embedder, device)
        with open(DB_PATH, "wb") as f:
            pickle.dump(database, f)
        print("✅ Built and saved authorized embeddings.")
    return database
```

```python
def recognize_person(img_path, database, mtcnn, embedder, device):
    img = Image.open(img_path).convert("RGB")
    face = mtcnn(img)

    if face is None:
        return "No Face Detected", None

    face = face.unsqueeze(0).to(device)

    with torch.no_grad():
        emb = embedder(face)
        emb = emb.cpu().numpy()

    min_dist = float("inf")
    identity = "Unknown"

    for name, db_emb in database.items():
        dist = np.linalg.norm(db_emb - emb)
        if dist < min_dist:
            min_dist = dist
            identity = name

    if min_dist < THRESHOLD:
        return f"✅ Authorized: {identity}", min_dist
    else:
        return "🚨 INTRUDER DETECTED", min_dist
```

```python
def main():
    mtcnn, embedder = initialize_models(device)

    database = load_or_create_database(mtcnn, embedder, device)

    # print("\n🔍 Running Test Recognition...")
    # result, distance = recognize_person(
    #     TEST_IMAGE_PATH, database, mtcnn, embedder, device
    # )
    while True:
        img_path = input("Image Path> ").strip()

        if img_path.lower() in ("q", "quit", "exit"):
            break

        if not os.path.isfile(img_path):
            print("[ERROR] Image file not found.")
            continue

        # label, dist = identify_person(img_path, database, INTRUDER_THRESHOLD)
        result, distance = recognize_person(
        img_path, database, mtcnn, embedder, device
        )
        # if result == "INTRUDER":
        #     print(f"ALERT! UNKNOWN PERSON DETECTED  (distance = {distance:.4f})\n")
        # else:
        #     print(f"✅ ACCESS GRANTED: {label}  (distance = {distance:.4f})\n")
        print("Result:", result)
        print("Distance:", distance)
```

```python
import os
import glob
import sys
from typing import Dict, Tuple, Optional

import torch
from torch import Tensor
from facenet_pytorch import MTCNN, InceptionResnetV1
from PIL import Image
import matplotlib.pyplot as plt


# ---------------------------
# CONFIG
# ---------------------------

# Folder of AUTHORIZED people:
# Each subfolder = one person
# Images inside are that person's face photos
DATA_ROOT = r"D:\MS-AI\fall-25\deepLearning\term_project\SecurityFaceDetection\data\lfw-deepfunneled"

# Threshold for intruder detection (tune later)
INTRUDER_THRESHOLD = 0.9

# Acceptable image types
IMG_EXTS = (".jpg", ".jpeg", ".png")

# ---------------------------
# DEVICE + MODELS (TRANSFER LEARNING)
# ---------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Face detector
mtcnn = MTCNN(image_size=160, margin=0, device=device)

# Pretrained FaceNet embedding model (TRANSFER LEARNING)
embedder = InceptionResnetV1(pretrained="vggface2").to(device).eval()


# ---------------------------
# EMBEDDING FUNCTIONS
# ---------------------------

def get_embedding(img_path: str) -> Optional[Tensor]:
    """
    Detect face and extract 512-D embedding.
    Returns a CPU tensor or None if face not detected.
    """
    try:
        img = Image.open(img_path).convert("RGB")
    except Exception as e:
        print(f"[WARN] Failed to open {img_path}: {e}")
        return None

    face = mtcnn(img)

    if face is None:
        print(f"[WARN] No face detected in {img_path}")
        return None

    face = face.unsqueeze(0).to(device)

    with torch.no_grad():
        emb = embedder(face)

    return emb.squeeze(0).cpu()


def build_database(root: str) -> Dict[str, Tensor]:
    """
    Build a database:
    person_name -> mean embedding of all their face images
    """
    if not os.path.isdir(root):
        print(f"[ERROR] DATA_ROOT not found: {root}")
        sys.exit(1)

    people = sorted(d for d in os.listdir(root)
                    if os.path.isdir(os.path.join(root, d)))

    database: Dict[str, Tensor] = {}
    print(f"\nBuilding authorized database from: {root}")

    for person in people:
        if person.startswith("A"):
            person_dir = os.path.join(root, person)
            img_paths = []
            for ext in IMG_EXTS:
                img_paths.extend(glob.glob(os.path.join(person_dir, f"*{ext}")))

            if not img_paths:
                continue

            embs = []
            for img_path in img_paths:
                emb = get_embedding(img_path)
                if emb is not None:
                    embs.append(emb)

            if not embs:
                continue

            mean_emb = torch.stack(embs).mean(dim=0)
            database[person] = mean_emb

            print(f"Loaded {person}: {len(embs)} images")

    print(f"\nAuthorized identities loaded: {len(database)}")
    return database


# ---------------------------
# IDENTIFICATION FUNCTION
# ---------------------------

def identify_person(
    img_path: str,
    database: Dict[str, Tensor],
    threshold: float
) -> Tuple[str, float]:
    """
    Compare input face to all authorized embeddings.
    Returns:
        (identity_label, distance)
    """

    emb = get_embedding(img_path)

    if emb is None:
        return "NO_FACE_DETECTED", float("inf")

    dists = []
    for name, ref_emb in database.items():
        dist = torch.norm(ref_emb - emb).item()
        dists.append((name, dist))

    best_name, best_dist = min(dists, key=lambda x: x[1])

    if best_dist > threshold:
        return "INTRUDER", best_dist
    else:
        return best_name, best_dist


# ---------------------------
# MAIN SECURITY LOOP
# ---------------------------

def main():
    print("\n=== FACE RECOGNITION SECURITY SYSTEM (TRANSFER LEARNING) ===")
    print(f"DATA_ROOT: {DATA_ROOT}")
    print(f"INTRUDER THRESHOLD: {INTRUDER_THRESHOLD}\n")

    database = build_database(DATA_ROOT)

    if not database:
        print("[ERROR] No authorized users found. Exiting.")
        return

    print("\nSystem Ready.")
    print("Enter an image path to test.")
    print("Type 'q' to quit.\n")

    while True:
        img_path = input("Image Path> ").strip()

        if img_path.lower() in ("q", "quit", "exit"):
            break

        if not os.path.isfile(img_path):
            print("[ERROR] Image file not found.")
            continue

        label, dist = identify_person(img_path, database, INTRUDER_THRESHOLD)

        if label == "INTRUDER":
            print(f"ALERT! UNKNOWN PERSON DETECTED  (distance = {dist:.4f})\n")
        else:
            print(f"✅ ACCESS GRANTED: {label}  (distance = {dist:.4f})\n")


if __name__ == "__main__":
    main()
```

```python
import os
import torch
import pickle
import numpy as np
from PIL import Image
from facenet_pytorch import MTCNN, InceptionResnetV1
import matplotlib.pyplot as plt

# ========================
# ✅ CONFIG
# ========================
AUTHORIZED_TEST = r"D:\\MS-AI\\fall-25\deepLearning\\term_project\\SecurityFaceDetection\\data\\lfw-deepfunneled"
INTRUDER_TEST = r"D:\\MS-AI\\fall-25\\deepLearning\\term_project\\SecurityFaceDetection\\test_data"
DB_PATH = "authorized_embeddings.pkl"

# ✅ Load threshold values to test
THRESHOLDS = np.linspace(0.4, 1.4, 50)

# ========================
# ✅ LOAD GPU + MODELS
# ========================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
mtcnn = MTCNN(image_size=160, margin=20, device=device)
embedder = InceptionResnetV1(pretrained="vggface2").to(device).eval()

with open(DB_PATH, "rb") as f:
    database = pickle.load(f)

# ========================
# ✅ EMBEDDING FUNCTION
# ========================
def get_embedding(img_path):
    # ✅ 1. Skip directories completely
    if not os.path.isfile(img_path):
        print(f"⚠️ Skipping non-file: {img_path}")
        return None

    # ✅ 2. Safely attempt to open the file as an image
    try:
        img = Image.open(img_path).convert("RGB")
    except PermissionError:
        print(f"⛔ Permission denied: {img_path}")
        return None
    except Exception as e:
        print(f"⚠️ Skipping invalid image: {img_path} → {e}")
        return None

    # ✅ 3. Face detection
    face = mtcnn(img)
    if face is None:
        print(f"⚠️ No face detected: {img_path}")
        return None

    # ✅ 4. Move to GPU/CPU
    face = face.unsqueeze(0).to(device)

    # ✅ 5. Generate embedding safely
    with torch.no_grad():
        emb = embedder(face)

    return emb.cpu().numpy()

# ========================
# ✅ DISTANCE COMPUTATION
# ========================
def compute_min_distance(emb, database):
    min_dist = float("inf")
    for db_emb in database.values():
        dist = np.linalg.norm(db_emb - emb)
        min_dist = min(min_dist, dist)
    return min_dist

# ========================
# ✅ RUN FULL EVALUATION
# ========================
authorized_distances = []
intruder_distances = []

# ✅ Authorized Tests
for root, _, files in os.walk(AUTHORIZED_TEST):
    for img_name in files:
        img_path = os.path.join(root, img_name)

        emb = get_embedding(img_path)
        if emb is None:
            continue

        dist = compute_min_distance(emb, database)
        authorized_distances.append(dist)

# ✅ Intruder Tests
for root, _, files in os.walk(INTRUDER_TEST):
    for img_name in files:
        img_path = os.path.join(root, img_name)

        emb = get_embedding(img_path)
        if emb is None:
            continue

        dist = compute_min_distance(emb, database)
        intruder_distances.append(dist)

# ========================
# ✅ FAR, FRR, TAR COMPUTATION
# ========================
FAR = []
FRR = []
TAR = []

for t in THRESHOLDS:
    FAR.append(np.mean(np.array(intruder_distances) < t))
    FRR.append(np.mean(np.array(authorized_distances) > t))
    TAR.append(1 - FRR[-1])

# ========================
# ✅ EER COMPUTATION
# ========================
FAR = np.array(FAR)
FRR = np.array(FRR)

eer_idx = np.argmin(np.abs(FAR - FRR))
EER = (FAR[eer_idx] + FRR[eer_idx]) / 2

print(f"✅ Equal Error Rate (EER): {EER:.4f}")

# ========================
# ✅ ROC CURVE
# ========================
plt.plot(FAR, TAR)
plt.xlabel("False Acceptance Rate (FAR)")
plt.ylabel("True Acceptance Rate (TAR)")
plt.title("ROC Curve - Face Recognition Security System")
plt.grid()
plt.show()
```

```python
import os
import glob
import pickle
import numpy as np
from PIL import Image
import torch
from facenet_pytorch import MTCNN, InceptionResnetV1
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report



#Authorized people
DATA_ROOT = r"data/lfw-deepfunneled"
#Custom faces       
CUSTOM_INTRUDER_ROOT = r"custom_test"         
DB_PATH = "authorized_embeddings.pkl"

THRESHOLD_SWEEP = np.linspace(0.4, 1.4, 50)
IMG_EXTENSIONS = (".jpg", ".jpeg", ".png")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"\nUsing device: {device}\n")
mtcnn = MTCNN(image_size=160, margin=20, device=device)
embedder = InceptionResnetV1(pretrained="vggface2").to(device).eval()

with open(DB_PATH, "rb") as f:
    database = pickle.load(f)

print(f"The number of loaded authorized identities: {len(database)}\n")

def get_embedding(img_path):
    if not os.path.isfile(img_path):
        return None
    try:
        img = Image.open(img_path).convert("RGB")
    except:
        return None
    face = mtcnn(img)
    if face is None:
        return None
    face = face.unsqueeze(0).to(device)
    with torch.no_grad():
        emb = embedder(face)
    return emb.cpu().numpy().squeeze(0)


def compute_min_dist(emb, database):
    min_dist = float("inf")
    for db_emb in database.values():
        dist = np.linalg.norm(db_emb - emb)
        if dist < min_dist:
            min_dist = dist
    return min_dist


authorized_distances = []
for root, _, files in os.walk(DATA_ROOT):
    for file in files:
        if file.lower().endswith(IMG_EXTENSIONS):
            path = os.path.join(root, file)
            emb = get_embedding(path)
            if emb is None:
                continue
            dist = compute_min_dist(emb, database)
            authorized_distances.append(dist)

print(f"Authorized test samples: {len(authorized_distances)}")


intruder_distances = []
for file in os.listdir(CUSTOM_INTRUDER_ROOT):
    if file.lower().endswith(IMG_EXTENSIONS):
        path = os.path.join(CUSTOM_INTRUDER_ROOT, file)
        emb = get_embedding(path)
        if emb is None:
            continue
        dist = compute_min_dist(emb, database)
        intruder_distances.append(dist)

print(f"Custom intruder test samples: {len(intruder_distances)}\n")


FAR = []   #Intruders that were incorrectly accepted
FRR = []   #Authorized users that were incorrectly rejected
TAR = []   #1 - FRR

authorized_distances = np.array(authorized_distances)
intruder_distances = np.array(intruder_distances)

for t in THRESHOLD_SWEEP:
    FAR.append(np.mean(intruder_distances < t))
    FRR.append(np.mean(authorized_distances > t))
    TAR.append(1 - FRR[-1])

FAR = np.array(FAR)
FRR = np.array(FRR)
TAR = np.array(TAR)


eer_idx = np.argmin(np.abs(FAR - FRR))
EER = (FAR[eer_idx] + FRR[eer_idx]) / 2
best_threshold = THRESHOLD_SWEEP[eer_idx]

print(f" Equal Error Rate (EER): {EER}")
print(f" Best Threshold: {best_threshold}")


#ROC curve plot
plt.plot(FAR, TAR)
plt.xlabel("False Acceptance Rate (FAR)")
plt.ylabel("True Acceptance Rate (TAR)")
plt.title("ROC Curve - Face Recognition Security")
plt.grid(True)
plt.show()


y_true = ([1] * len(authorized_distances) + [0] * len(intruder_distances))
y_pred = ([1 if d < best_threshold else 0 for d in authorized_distances] + [1 if d < best_threshold else 0 for d in intruder_distances])

cm = confusion_matrix(y_true, y_pred)
print("\nConfusion Matrix (Authorized=1, Intruder=0):")
print(cm)
print("\nClassification Report:")
print(classification_report(y_true, y_pred))

def test_single_image(img_path, threshold=best_threshold):
    emb = get_embedding(img_path)
    if emb is None:
        print("No face has been detected.")
        return

    dist = compute_min_dist(emb, database)
    print(f"Distance = {dist}")
    if dist < threshold:
        print("Authorized")
    else:
        print("Intruder")
```
